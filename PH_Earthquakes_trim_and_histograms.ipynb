# Magnitude Histogram

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('phivolcs_earthquake_oct.csv')

df.head()
```

```python
print(df.shape)
```

```python
# CLEAN NUMERIC COLUMNS

# Replace hyphens with NaN
df = df.replace('-', np.nan)

# Convert numeric columns properly
numeric_cols = ['Latitude', 'Longitude', 'Depth_In_Km', 'Magnitude']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Drop rows that still contain NaN in these fields
df = df.dropna(subset=numeric_cols)

print(df.shape)
```

```python
print(df.shape)
```

```python
df.head()
```

```python
import matplotlib.pyplot as plt

labels = ['≤2.5', '2.5–5.4', '5.5–6.0', '6.1–6.9', '7.0–7.9', '≥8.0']

df['Mag_Bin'] = pd.cut(
  df['Magnitude'],
  bins=[-float('inf'), 2.5, 5.4, 6.0, 6.9, 7.9, float('inf')],
  labels=labels)

counts = df['Mag_Bin'].value_counts().sort_index()

counts.plot(kind='bar', edgecolor='black')

plt.xlabel("Magnitude Range", fontname='DejaVu Serif')
plt.ylabel("Frequency", fontname='DejaVu Serif')
plt.title("Earthquake Magnitude Distribution", fontname='DejaVu Serif')

plt.show()
```

```python
import matplotlib.pyplot as plt

# Define labels
labels = ['≤2.5', '2.5–5.4', '5.5–6.0', '6.1–6.9', '7.0–7.9', '≥8.0']

# Define bins (aligned with your labels)
bins = [-float('inf'), 2.5, 5.4, 6.0, 6.9, 7.9, float('inf')]

# Categorize
df['Mag_Bin'] = pd.cut(df['Magnitude'], bins=bins, labels=labels)

# Count + convert to percentages
counts = df['Mag_Bin'].value_counts().sort_index()
percentages = counts / counts.sum() * 100

# Print percentages
print(percentages)
```

```python
import numpy as np
import matplotlib.pyplot as plt

magnitudes = df['Magnitude']

# Create bins every 0.1
bins = np.arange(magnitudes.min(), magnitudes.max() + 0.1, 0.1)

weights = np.ones_like(magnitudes) / len(magnitudes) * 100

plt.hist(magnitudes, bins=bins, weights=weights, edgecolor='navy')

plt.xlabel("Magnitude", fontname='DejaVu Serif')
plt.ylabel("Percentage (%)", fontname='DejaVu Serif')
plt.title("Earthquake Magnitude Distribution (0.1 bins)", fontname='DejaVu Serif')

plt.show()
```
